# Nigeria soil data explorer

Four sources, one dataframe each, all tagged with the same `state` and `geopolitical_zone` columns.

```python
import pandas as pd

from nasa import SoilExplorer
from nasa import Climate

# pd.set_option("display.max_columns", 40)
pd.set_option('display.max_columns', None)
pd.set_option("display.width", 200)
ex = SoilExplorer()
clim = Climate()
lat, lon = 9.0765, 7.3986
```

## Overview of the three file sources

The first run builds each table and saves it in the database, later runs read it from there.

```python
ex.overview()
```

## 1. AfSP legacy soil profiles (ISRIC)

```python
afsp = ex.afsp.frame()
afsp.head()
```

```python
afsp.groupby(['geopolitical_zone', 'state']).size().rename('layers')
```

## 2. iSDA soil samples (2008 to 2018)

```python
samples = ex.samples.frame()
samples.head()
```

```python
samples.groupby(['geopolitical_zone', 'state']).size().rename('samples')
```

## 3. Nigeria soil map (1985 to 1990 survey)

One row per map polygon per state. `piece_share` is the part of the polygon that falls in that state.

```python
soil_map = ex.soil_map.frame()
soil_map.drop(columns='geometry').head()
```

```python
soil_map.plot(column='drainage', legend=True, figsize=(8, 8))
```

## 4. iSDAsoil API, one coordinate

Needs `ISDA_USERNAME` and `ISDA_PASSWORD` in the `.env` file.

```python

api = ex.api.point(lat, lon)
api.drop(columns=['lat', 'lon']).head(10)
```

```python
print(len(api), 'values,', len(api.attrs['failed']), 'failed requests')
api.attrs['failed']
```

```python
w = clim.power.series(lat, lon)
w.head()
```

```python
print(w.attrs)
clim.power.coverage()
```

```python
ts = w.set_index("date")
ts[["T2M_MAX", "T2M_MIN"]].resample("MS").mean().plot(figsize=(10, 4))
```
